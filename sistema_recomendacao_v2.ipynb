```python
## 4. Refinamento do Modelo (Etapa 3)
### 4.1 Análise dos Resultados Preliminares
Na prova de conceito (V1), utilizamos o `MinMaxScaler` e a métrica de `cosseno`. Observou-se que variáveis com muitos *outliers* (como *loudness* e *tempo*) poderiam distorcer ligeiramente a similaridade.

### 4.2 Ajuste do Pipeline de Treinamento
Para melhorar o desempenho, faremos duas alterações simples no pipeline:
1. Trocar o `MinMaxScaler` pelo `StandardScaler` para mitigar o efeito de outliers.
2. Testar a métrica `euclidean` (Euclidiana) em comparação ao `cosine` (Cosseno) e ajustar o número de vizinhos ($k=10$).
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import NearestNeighbors
import pandas as pd

# 1. Preparação com StandardScaler (Substituindo MinMaxScaler)
scaler_v2 = StandardScaler()
features = ['danceability', 'energy', 'valence', 'acousticness', 'loudness', 'tempo']
X_scaled_v2 = scaler_v2.fit_transform(df[features])

# 2. Novo Treinamento (Ajuste de hiperparâmetros)
# Testando a métrica euclidiana e k=10
modelo_v2 = NearestNeighbors(n_neighbors=10, metric='euclidean', algorithm='brute')
modelo_v2.fit(X_scaled_v2)

# 3. Função de Recomendação V2
def recomendar_musicas_v2(nome_musica, df, modelo, scaler):
    idx = df[df['track_name'].str.contains(nome_musica, case=False, na=False)].index[0]
    vetor_musica = scaler.transform(df.loc[[idx], features])

    distancias, indices = modelo.kneighbors(vetor_musica)

    recomendacoes = df.iloc[indices[0]]
    return recomendacoes[['track_name', 'artists', 'track_genre']]

# Testando o modelo
recomendar_musicas_v2("Nome de uma Música Teste", df, modelo_v2, scaler_v2)
```
